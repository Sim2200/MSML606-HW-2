## Part 1 - Stacks and Binary Trees

```python
import csv
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class HomeWork2:

    # Problem 1: Construct an expression tree (Binary Tree) from a postfix expression

    def constructBinaryTree(self, input) -> TreeNode:      
        stack = []
        operators = '+-*/'

        for token in input:
            if token not in operators:
                # push as tree node
                stack.append(TreeNode(token))
            else:
                if len(stack) < 2:
                    raise ValueError(f"Invalid postfix expression: not enough operands for '{token}'")
                # now put left and right nodes
                right = stack.pop()
                left = stack.pop()
                node = TreeNode(token, left, right)
                stack.append(node)

        return stack[0] if stack else None


    # Problem 2.1: Use pre-order traversal (root, left, right) to generate prefix notation

    def prefixNotationPrint(self, head: TreeNode) -> list:
        result = []
        def preorder(node):
            if node:   
                result.append(node.val) # see root
                preorder(node.left)   # see left subtree
                preorder(node.right) # see right subtree

        preorder(head)
        return result

    # Problem 2.2: Use in-order traversal (left, root, right) for infix notation with appropriate parentheses.
  
    def infixNotationPrint(self, head: TreeNode) -> list:
        if not head:
            return []

        result = []  

        def inorder(node):
            if node:
                if node.val in ['+', '-', '*', '/']:
                    result.append('(')
                inorder(node.left)  # see left subtree
                result.append(node.val)   # see root
                inorder(node.right) # see right subtree
                if node.val in ['+', '-', '*', '/']:
                    result.append(')')

        inorder(head)
        return result


    # Problem 2.3: Use post-order traversal (left, right, root) to generate postfix notation.

    def postfixNotationPrint(self, head: TreeNode) -> list:
        result = []

        def postorder(node):
            if node:
                postorder(node.left)  # see left subtree
                postorder(node.right)  # see right subtree
                result.append(node.val)  # see root

        postorder(head)
        return result


class Stack:
    # Implement your stack using either an array or a list

    def __init__(self):
        self.items = []
    
    # check if stack is empty
    def is_empty(self):
        return len(self.items) == 0
    
    # get size of stack
    def size(self):
        return len(self.items)

    # push an element to top of stack
    def push(self, val):
       self.items.append(val)

    # remove/pop elemnet from top of stack
    def pop(self):
        if self.is_empty():
            raise IndexError("Stack is empty, can't pop")
        return self.items.pop()

    # see top element of stack
    def top(self):
        if self.is_empty():
            raise IndexError("Stack is empty")
        return self.items[-1]



    # Problem 3: Write code to evaluate a postfix expression using stack and return the integer value
    
    def evaluatePostfix(self, exp: str) -> int:
        operators = '+-*/'
        for token in exp.split():
            if token in operators:
                if self.size() < 2:
                    raise ValueError("Invalid postfix expression: insufficient operands")
                b = self.pop()
                a = self.pop()

                if token == '+':
                    self.push(a + b)
                elif token == '-':
                    self.push(a - b)
                elif token == '*':
                    self.push(a * b)
                elif token == '/':
                    if b == 0:
                        raise ZeroDivisionError("division by zero")
                    self.push(int(a / b))  # integer division
            else:
                try:
                    self.push(int(token)) # push
                except ValueError:
                    raise ValueError(f"Invalid token in postfix expression: '{token}'")
        return self.pop()

```

## Problem 4

#### Explain how edge cases are handled.

Edge cases are handled by explicitly testing for inputs at extreme cases.

1. Empty postfix expressions - If the input string is empty or contains only blank spaces, a ValueError is raised with the message "Empty postfix expression".
2. Malformed postfix expressions - Here, the size of the stack is checked, If there are not enough operands, then a ValueError is raied. After running, if more than one operand remains, ValueError is raised for leftover operands
3. Division by zero - If divisor (here b) is 0 in division, a ZeroDivisionError is raised.
4. Invalid tokens - If a token is not one of the operators (+,-,/,*) or cannot be converted to an integer, then a ValueError is raised.
5. Very large numbers or results - Python integers handle arbitrary large numbers, so operations work without overflow.
6. Negavtive numbers in expression - Negative numbers are supported naturally by int(token) conversion, so arithmetic works as expected.

### Example test cases showing edge case handling

```python
s = Stack()

# 1. Empty expression
try:
    s.evaluatePostfix("")
except ValueError as e:
    print(e)  # Output: Empty postfix expression

# 2. Insufficient operands
try:
    s.evaluatePostfix("5 +")
except ValueError as e:
    print(e)  # Output: Malformed postfix expression: insufficient operands

# 3. Extra operands
try:
    s.evaluatePostfix("5 3 2 +")
except ValueError as e:
    print(e)  # Output: Malformed postfix expression: leftover operands

# 4. Division by zero
try:
    s.evaluatePostfix("10 0 /")
except ZeroDivisionError as e:
    print(e)  # Output: division by zero

# 5. Invalid token
try:
    s.evaluatePostfix("10 a +")
except ValueError as e:
    print(e)  # Output: Invalid token in postfix expression

# 6. Negative numbers
print(s.evaluatePostfix("-3 -2 *"))  # Output: 6

# 7. Very large numbers
print(s.evaluatePostfix("1000000000 3000000000 +"))  # Output: 4000000000
```

### Approach 

#### Problem 1 -

Stack-based construction from postfix:
- Operands - push as TreeNode.
- Operators - pop two nodes, make them left/right children, push back as a new node.

Edge cases handled:
- Empty postfix - stack ends up empty - None returned.
- Malformed postfix - popping from empty stack raises IndexError 

#### Problem 2 -
Prefix (pre-order): root → left → right
- Traversal stores node values in a list.
Infix (in-order with parentheses): left → root → right
- Wrap operator nodes in parentheses to clarify operation order.
- Edge case: single-node tree - outer parentheses still included
Postfix (post-order): left → right → root
- Traversal stores node values in a list.


#### Problem 3 -

Postfix Evaluation: 
- Use a stack to store operands.
- Iterate through tokens - If token is a number, push onto the stack & If token is an operator, pop two operands, perform the operation, push the result.

Edge cases:
- Empty input: raise ValueError.
- Insufficient operands - ValueError.
- Malformed input: check stack size before popping; ensure only one operand remains at end.
- Division by zero: check divisor before division, raise ZeroDivisionError.
- Invalid token: non-integer or unsupported operator raises ValueError.

Return the final result from the stack.

This approach ensures safe handling of invalid expressions, prevents crashes, and handles both negative and large numbers.
